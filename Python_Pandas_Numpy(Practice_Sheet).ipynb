```python
# Question 35.
#  Examination performance analysis: Generate a reproducible 12 x 4 NumPy array of integer test marks from 20 to 100, inclusive, using a fixed seed. Calculate each student's average and each test's mean and standard deviation. Count students passing all four tests (at least 40 in each). Identify students with averages of at least 75 and the test with greatest variation. State your student numbering convention.


import numpy as np

np.random.seed(42)

marks = np.random.randint(20, 101, size=(12, 4))

student_average = marks.mean(axis=1)

test_mean = marks.mean(axis=0)
test_std = marks.std(axis=0)

passing_students = np.all(marks >= 40, axis=1)
pass_count = np.sum(passing_students)

high_average_students = np.where(student_average >= 75)[0] + 1

greatest_variation_test = np.argmax(test_std) + 1

print("Marks:\n", marks)
print("Student averages:", student_average)
print("Test means:", test_mean)
print("Test standard deviations:", test_std)
print("Students passing all tests:", pass_count)
print("Students with average >= 75:", high_average_students)
print("Test with greatest variation: Test", greatest_variation_test)
```

```python
# Question 36.
# Matrix transformation and analysis: Construct a 5 x 5 matrix containing integers 21 to 45. Extract the central 3 x 3 matrix, transpose it, and calculate the matrix product of the extracted matrix with its transpose. Calculate row-wise sums of the product and identify its maximum and all positions tied at that maximum. Explain dimensions at every stage.

import numpy as np

matrix = np.arange(21, 46).reshape(5, 5)

central = matrix[1:4, 1:4]

transpose = central.T

product = np.matmul(central, transpose)

row_sums = product.sum(axis=1)

maximum = np.max(row_sums)

positions = np.where(row_sums == maximum)[0]

print("Original Matrix:\n", matrix)
print("Shape:", matrix.shape)

print("\nCentral 3x3 Matrix:\n", central)
print("Shape:", central.shape)

print("\nTranspose:\n", transpose)
print("Shape:", transpose.shape)

print("\nProduct Matrix:\n", product)
print("Shape:", product.shape)

print("\nRow-wise sums:", row_sums)
print("Maximum row sum:", maximum)
print("Positions with maximum:", positions)
```

```python
# Question 37.
# Sales and revenue analysis: Create a 6 x 4 matrix of quantities sold over six days for four products. Use unit prices [150, 240, 90, 320]. Calculate daily revenue, total revenue per product, the highest-revenue day, and the best-selling product by quantity. Reduce prices by 10% for products whose total quantity exceeds 100, then recalculate total revenue without modifying original prices.

import numpy as np

quantity = np.array([
    [20, 15, 30, 10],
    [25, 18, 22, 15],
    [30, 25, 20, 18],
    [15, 30, 28, 20],
    [35, 22, 25, 25],
    [18, 35, 32, 30]
])

prices = np.array([150, 240, 90, 320])

daily_revenue = quantity @ prices

product_revenue = quantity.sum(axis=0) * prices

highest_revenue_day = np.argmax(daily_revenue) + 1

total_quantity = quantity.sum(axis=0)
best_selling_product = np.argmax(total_quantity) + 1

new_prices = np.where(total_quantity > 100, prices * 0.9, prices)

new_product_revenue = total_quantity * new_prices

print("Daily revenue:", daily_revenue)
print("Total revenue per product:", product_revenue)
print("Highest revenue day: Day", highest_revenue_day)
print("Total quantity per product:", total_quantity)
print("Best-selling product: Product", best_selling_product)
print("Reduced prices:", new_prices)
print("New total revenue per product:", new_product_revenue)
```

```python
# Question 38.
# Random data cleaning and statistics: Generate 100 normal observations with mean 50 and standard deviation 12 using a fixed seed. Count values outside 30-70 and replace them with the nearest boundary using np.where(). Compare original and cleaned mean, variance, standard deviation, minimum, and maximum. Explain the statistical effects of boundary replacement.
import numpy as np

np.random.seed(42)
data = np.random.normal(loc=50, scale=12, size=100)

outside_count = np.sum((data < 30) | (data > 70))

cleaned_data = np.where(data < 30, 30,
                np.where(data > 70, 70, data))

def statistics(arr):
    return {
        "Mean": np.mean(arr),
        "Variance": np.var(arr),
        "Standard Deviation": np.std(arr),
        "Minimum": np.min(arr),
        "Maximum": np.max(arr)
    }

original_stats = statistics(data)
cleaned_stats = statistics(cleaned_data)

print("Values outside 30-70:", outside_count)

print("\nOriginal Data Statistics")
for key, value in original_stats.items():
    print(f"{key}: {value:.2f}")

print("\nCleaned Data Statistics")
for key, value in cleaned_stats.items():
    print(f"{key}: {value:.2f}")

```

```python
# Multi-branch financial reconciliation: Combine the three earnings Series by ordinary addition. Explain missing values, identify quarters common to all branches, and calculate their combined earnings. Construct a separate result treating absent branch records as zero and explain how its interpretation differs. branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180}) branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155}) branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})



import pandas as pd

branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

combined = branch_a + branch_b + branch_c
print("Ordinary Addition:")
print(combined)

common_quarters = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print("\nCommon Quarters:")
print(common_quarters)

print("\nCombined Earnings (Common Quarters):")
print(combined.loc[common_quarters])

combined_zero = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print("\nAddition with Missing Values Treated as Zero:")
print(combined_zero)
```

```python
# Series data integrity: Preserve the required index order, identify missing student records, and filter valid scores of at least 70. Convert the filtered Series to a dictionary, reconstruct it as a Series, and verify that labelled values match. Explain why missing records should not automatically be zero marks. scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48} required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102'] s = pd.Series(scores, index=required)

import pandas as pd

scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

s = pd.Series(scores, index=required)

print("Series:")
print(s)

missing = s[s.isna()]
print("\nMissing Student Records:")
print(missing)

filtered = s[s >= 70]
print("\nValid Scores (>=70):")
print(filtered)

score_dict = filtered.to_dict()
print("\nDictionary:")
print(score_dict)

new_series = pd.Series(score_dict)
print("\nReconstructed Series:")
print(new_series)

print("\nValues Match:")
print(filtered.equals(new_series))
```
