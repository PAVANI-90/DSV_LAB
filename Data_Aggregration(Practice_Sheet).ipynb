```python
# Ques 12.

import pandas as pd

df = pd.read_csv("Retail_Data_Science_Dataset.csv")

result = df.groupby(["branch", "product"], as_index=False)["units_sold"].mean()

print(result)
```

```python
# Ques 13.
result = df.groupby(["branch", "product"], as_index=False)["discount_pct"].mean()

print(result)
```

```python
# Ques 14.
result = df.groupby(["branch", "product"], as_index=False)["units_sold"].max()
print(result)
```

```python
# Ques 15.
result = df.groupby(["branch", "product"], as_index=False)["customer_rating"].mean()
print(result)
```

```python
# Ques 16.
result = df.groupby("branch", as_index=False).agg(
    total_units_sold=("units_sold", "sum"),
    average_units_sold=("units_sold", "mean"),
    minimum_units_sold=("units_sold", "min"),
    maximum_units_sold=("units_sold", "max"),
    standard_deviation=("units_sold", "std")
)
print(result)
```

```python
# Ques 17.
result = df.groupby("product", as_index=False).agg(
    total_units_sold=("units_sold", "sum"),
    average_discount_pct=("discount_pct", "mean"),
    maximum_customer_rating=("customer_rating", "max"),
    minimum_delivery_time=("delivery_minutes", "min")
)
print(result)
```

```python
# Ques 18.
result = df.groupby("branch").agg(
    total_sales_records=("branch", "size"),
    non_missing_customer_ratings=("customer_rating", "count")
).reset_index()
print(result)
```

```python
# Ques 19.
def find_range(x):
    return x.max() - x.min()
result = df.groupby("product", as_index=False).agg(
    units_sold_range=("units_sold", find_range)
)
print(result)
```

```python
# Ques 20.
result = df.groupby(["branch", "product"], as_index=False).agg(total_units_sold=("units_sold", "sum"),total_target_units=("target_units", "sum"))
print(result)
```

```python
# Ques 21.
grouped = df.groupby("branch")
for branch, group in grouped:
    print("Branch:", branch)
    print(group)
    print("Total Units Sold:", group["units_sold"].sum())
    print("-" * 50)
```

```python
# Ques 22.
result = df.groupby("product")["units_sold"].quantile([0.25, 0.50, 0.75]).unstack()
result.columns = ["25th Percentile", "50th Percentile", "75th Percentile"]
print(result)
```

```python
# Ques 23.
df_hierarchical = df.set_index(["branch", "product"])
result = df_hierarchical.groupby(level="branch")["units_sold"].sum()
print(result)
```

```python
# Ques 24.
result = pd.pivot_table(df,values="units_sold",index="branch",columns="product",aggfunc="sum",margins=True,margins_name="Total")
print(result)
```

```python
# Ques 25.
result = pd.pivot_table(df,values=["discount_pct", "customer_rating"],index="branch",columns="product",aggfunc="mean")
print(result)
```

```python
# Ques 26.
print("Before Replacement:")
print(df)
df["customer_rating"] = df["customer_rating"].fillna(df.groupby(["branch", "product"])["customer_rating"].transform("mean"))
print("After Replacement:")
print(df)
```

```python
# Ques 27.
def weighted_avg(group):
    return (group["units_sold"] * group["discount_pct"]).sum() / group["units_sold"].sum()
result = df.groupby("branch").apply(weighted_avg).reset_index(name="weighted_average_discount")
print(result)
```

```python
# Ques 28.
result = df.groupby("branch", group_keys=False).apply(lambda x: x.nlargest(3, "units_sold"))[["branch", "product", "day", "units_sold"]]
print(result)
```

```python
# Ques 29.
df["sales_contribution_pct"] = (df["units_sold"] /df.groupby("branch")["units_sold"].transform("sum")) * 100
print(df)
result = df.groupby("branch")["sales_contribution_pct"].sum()
print(result)
```

```python
# Ques 30.
df["Sales_Category"] = pd.qcut(df["units_sold"],q=4,labels=["Low Sales","Below Average Sales","Above Average Sales","High Sales"])
result = df.groupby("Sales_Category", as_index=False).agg(
    total_units_sold=("units_sold", "sum"),
    average_discount_pct=("discount_pct", "mean"),
average_customer_rating=("customer_rating", "mean"),average_delivery_time=("delivery_minutes", "mean"),number_of_sales_records=("units_sold", "count"))
print(result)
```
