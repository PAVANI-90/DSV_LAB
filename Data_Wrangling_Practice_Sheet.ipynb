```python
# Ques 27.

import pandas as pd

df1 = pd.read_csv("marks_original.csv")
df2 = pd.read_csv("marks_updated.csv")
df3 = pd.read_csv("students_master.csv")

df1 = df1.set_index("Student_ID")
df2 = df2.set_index("Student_ID")

marks = df1.combine_first(df2)

marks = marks.reset_index()

result = pd.merge(df3, marks, on="Student_ID", how="left")

print(result)
```

```python
# Ques 28.

import pandas as pd
import numpy as np

df1 = pd.read_csv("marks_original.csv")
df2 = pd.read_csv("marks_updated.csv")

df1 = df1.set_index("Student_ID")
df2 = df2.set_index("Student_ID")

combine_first_result = df1.combine_first(df2)

fmax_result = df1.combine(df2, np.fmax)

result = pd.DataFrame({
    "combine_first": combine_first_result["Marks"],
    "fmax": fmax_result["Marks"]
})

print(result)

print(result[result["combine_first"] != result["fmax"]])
```

```python
# Ques 29.

import pandas as pd

df = pd.read_csv("repeat_tests.csv")

result = df.pivot_table(
    index="Student_ID",
    columns="Subject",
    values="Marks",
    aggfunc=["mean", "max"]
)

print(result)

stacked = result.stack()

print(stacked)

restored = stacked.unstack()

print(restored)
```
